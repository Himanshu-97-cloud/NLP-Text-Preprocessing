Removing punctuation

🧹 Token-Level Processing

```python
# Importing libraries
import numpy as np
import pandas as pd
import nltk
import re
import contractions
```

**Text Cleaning and Normalization**

```python
# Text and Dataframe
text = '''Hello!!! My name's Himanshu, and I'm learning NLP 😊.  
I have 2 cats, 3 dogs, and 1 parrot.  
Sometimes I write like THIS, sometimes like this...  
BTW, check out my blog at https://myblog.com or email me at test_user123@example.com.  
Don't forget to follow me @coolguy and use #MachineLearning #DataScience.  
Prices went up by 50% in 2025, but now they're stable.  
Ummm...   too   many   spaces   here.  
'''

data = {
    "Text1": ["Hello world!!!", "Natural Language 2025", "Data   Science", "Machine-Learning 😊"],
    "Text2": ["Regex basics...", "Tokenization @user", "Preprocessing #NLP studies", "Embeddings https://example.com"],
    "Number": [10, 20, 30, 40]
}

df = pd.DataFrame(data)
df
```

`isinstance(object, classinfo)`

object → the variable or value you want to check.

classinfo → the type(s) you want to test against.

```python
# Lowercasing text
text = text.lower()
print(text)

# Lowercasing DataFrame
df = df.map(lambda s : s.lower() if isinstance(s, str) else s)
df
```

```python
# Handling Contractions in text
text = contractions.fix(text)
print(text)

# Handling contractions in DataFrame
def handle_contraction(text):
    return contractions.fix(text)
df['Text1'] = df['Text1'].apply(handle_contraction)
df['Text2'] = df['Text2'].apply(handle_contraction)
df
```

`re.sub(pattern, repl, string, count=0, flags=0)`

```python
# Removing Punctuations from text
text = re.sub(r"[^\w\s]", "", text)
print(text)

# Removing Punctuations from DataFrame
def remove_punctuation(text):
    return re.sub(r"[^\w\s]", "", text)
df['Text1'] = df['Text1'].apply(remove_punctuation)
df['Text2'] = df['Text2'].apply(remove_punctuation)
df
```

```python
# Removing numbers from text
text = re.sub(r"\d+", "", text)
print(text)

# Removing numbers from DataFrame
def remove_num(text):
    return re.sub(r"\d+", "", text)
df['Text1'] = df['Text1'].apply(remove_num)
df['Text2'] = df['Text2'].apply(remove_num)
df
```

```python
# Remove white spaces from Text
text = re.sub(r"\s", " ", text)
print(text)

# Removing white spaces from DstaFrame
def remove_spaces(text):
    return re.sub(r"\s", " ", text)
df['Text1'] = df['Text1'].apply(remove_spaces)
df['Text2'] = df['Text2'].apply(remove_spaces)
df
```

**Token Level Processing**

```python
from nltk.tokenize import word_tokenize
nltk.download('punkt')

# TOkenization of text
text = word_tokenize(text)
print(text)

# TOkenization of DataFrame
df['Text1'] = df['Text1'].apply(word_tokenize)
df['Text2'] = df['Text2'].apply(word_tokenize)
df
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')

# Lemmatization of Text 
lemmatizer = WordNetLemmatizer()
text = [lemmatizer.lemmatize(w) for w in text]
print(text)

# lemmatization in DataFrame 
def lemmatization(text):
    return [lemmatizer.lemmatize(w) for w in text]
df['Text1'] = df['Text1'].apply(lemmatization)
df['Text2'] = df['Text2'].apply(lemmatization)
df
```

```python
from nltk.stem import PorterStemmer

# Stemming of Text
stemmer = PorterStemmer()
text = [stemmer.stem(w) for w in text]
print(text)

# Stemming in DataFrame
def stemm(text):
    return [stemmer.stem(w) for w in text]
df['Text1'] = df['Text1'].apply(stemm)
df['Text2'] = df['Text2'].apply(stemm)
df
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')

stopword = stopwords.words('English')
text = [w for w in text if w not in stopword]
print(text)

def remove_stopwords(text):
    return [w for w in text if w not in stopword]
df['Text1'] = df['Text1'].apply(remove_stopwords)
df['Text2'] = df['Text2'].apply(remove_stopwords)
df
```
